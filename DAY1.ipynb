## 4. Axis Labels

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5, 6]
sales = [10000, 15000, 12000, 20000, 25000, 30000]

plt.plot(months, sales)

plt.title("Monthly Sales Report")

plt.xlabel("Months")
plt.ylabel("Sales (₹)")

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = [1, 2, 3, 4, 5]
marks = [65, 75, 80, 55, 90]

plt.plot(subjects, marks)

plt.title("Student Performance")

plt.xlabel("Subjects" , fontsize=14)
plt.ylabel("Marks" , fontsize=14)

plt.show()
```

## 5. Legends

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]

actual_sales = [100, 150, 120, 200, 250]
predicted_sales = [90, 140, 130, 190, 240]

plt.plot(months, actual_sales, label="Actual Sales")
plt.plot(months, predicted_sales, label="Predicted Sales")

plt.title("Actual vs Predicted Sales")

plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend(loc="upper right")

plt.show()
```
