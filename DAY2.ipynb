# Day 2 - Bar Charts and Histograms

## 1. Bar Charts

```python
import matplotlib.pyplot as plt
students = ["A", "B", "C", "D", "E"]
marks = [70, 85, 60, 90, 75]

plt.bar(students, marks)
plt.show()
```

```python
import matplotlib.pyplot as plt
students = ["A", "B", "C", "D", "E"]
marks = [70, 85, 60, 90, 75]
plt.bar(students, marks, color="skyblue")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
students = ["A", "B", "C", "D", "E"]
marks = [70, 85, 60, 90, 75]
plt.barh(students, marks, color="skyblue")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
products = ["Laptop", "Mobile", "Tablet", "Watch", "Camera"]

sales = [50, 80, 40, 65, 30]
plt.bar(products, sales, color = "purple")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [35, 40, 45, 50, 55, 60, 65, 70, 75, 80,
         42, 48, 53, 58, 62, 68, 72, 78, 85, 90]

plt.hist(marks)

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [35, 40, 45, 50, 55, 60, 65, 70, 75, 80,
         42, 48, 53, 58, 62, 68, 72, 78, 85, 90]

plt.hist(marks, bins=5)

plt.show()
```

```python

import matplotlib.pyplot as plt

marks = [35, 40, 45, 50, 55, 60, 65, 70, 75, 80,
         42, 48, 53, 58, 62, 68, 72, 78, 85, 90]

plt.hist(marks, bins=10)

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [35, 40, 45, 50, 55, 60, 65, 70, 75, 80,
         42, 48, 53, 58, 62, 68, 72, 78, 85, 90]
plt.hist(marks, bins=5, color="skyblue", edgecolor="black")

plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
import matplotlib.pyplot as plt
ages = [18, 19, 20, 21, 22, 18, 19, 20, 21, 22,
        18, 20, 21, 23, 24, 19, 22, 23, 20, 21]
plt.hist(ages, bins=5, color="orange", edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```
