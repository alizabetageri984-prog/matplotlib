# Day 3 - Subplots and Saving Figures

## 1. Subplots

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [10, 20, 15, 30, 40]
expenses = [5, 10, 12, 15, 20]

# First graph
plt.subplot(1, 2, 1)
plt.plot(months, sales)
plt.title("Monthly Sales")

# Second graph
plt.subplot(1, 2, 2)
plt.plot(months, expenses)
plt.title("Monthly Expenses")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [10, 20, 15, 30, 40]
expenses = [5, 10, 12, 15, 20]

fig, ax = plt.subplots(1, 2)

ax[0].plot(months, sales)
ax[0].set_title("Monthly Sales")

ax[1].plot(months, expenses)
ax[1].set_title("Monthly Expenses")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5, 6]

sales = [10000, 15000, 12000, 20000, 25000, 30000]

expenses = [5000, 8000, 7000, 10000, 12000, 15000]

fig, ax = plt.subplots(1, 2)

ax[0].plot(months, sales)
ax[0].set_title("Monthly Sales")
ax[0].set_xlabel("Months")
ax[0].set_ylabel("Sales")

ax[1].plot(months, expenses)
ax[1].set_title("Monthly Expenses")
ax[1].set_xlabel("Months")
ax[1].set_ylabel("Expenses")

plt.tight_layout()
plt.show()
```

Saving Figures

```python
import matplotlib.pyplot as plt
months = [1, 2, 3, 4, 5]

sales = [100, 150, 120, 200, 250]
plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.savefig("monthly_sales.png", dpi=300)
plt.savefig("monthly_sales.pdf")

plt.show()
```

```python
import matplotlib.pyplot as plt
months = [1, 2, 3, 4, 5]

sales = [100, 150, 120, 200, 250]
plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.savefig(
    "monthly_sales.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
