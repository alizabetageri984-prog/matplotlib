Box Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = sns.load_dataset("tips")

print(df.head())
```

```python
Q1= df["total_bill"].quantile(0.25)

Q1 = df["total_bill"].quantile(0.25)

Q3 = df["total_bill"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Boundary:", lower)
print("Upper Boundary:", upper)
```

```python
outliers = df[
    (df["total_bill"] < lower) |
    (df["total_bill"] > upper)
]

print("Number of outliers:", len(outliers))

print(outliers[["total_bill"]])
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, y="total_bill")

plt.title("Total Bill - Outlier Detection")
plt.ylabel("Total Bill")

plt.show()
```

Heatmaps

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = sns.load_dataset("tips")

print(df.head())
```

```python
numerical_data = df[["total_bill", "tip", "size"]]

print(numerical_data.head())
```

```python
correlation = numerical_data.corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(correlation)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    center=0,
    vmin=-1,
    vmax=1,
    cbar=True
)

plt.title("Professional Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="viridis",
    linewidths=0.5,
    center=0,
    vmin=-1,
    vmax=1,
    cbar=True
)

plt.title("Tips Dataset Correlation Heatmap")

plt.show()
```
