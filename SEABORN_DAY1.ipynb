```python
import seaborn as sns
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [35, 40, 50, 60, 75, 90]

sns.scatterplot(
    x=hours,
    y=marks,
    color="green",
    s=100
)

plt.title("Hours Studied vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```

```python
df = sns.load_dataset("tips")
```

```python
sns.countplot(data=df, x="day")

plt.title("Number of Customers by Day")
plt.xlabel("Day")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.countplot(
    data=df,
    x="day",
    palette="viridis"
)

plt.title("Number of Customers by Day")
plt.xlabel("Day")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.barplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.title("Average Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")

plt.show()
```

```python
sns.barplot(
    data=df,
    x="day",
    y="total_bill",
    errorbar=None
)

plt.title("Average Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")

plt.show()
```

```python
sns.histplot(
    data=df,
    x="total_bill"
)

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.histplot(
    data=df,
    x="total_bill"
)

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.histplot(
    data=df,
    x="total_bill",
    bins=5
)

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.histplot(
    data=df,
    x="total_bill",
    bins=20
)

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Number of Customers")

plt.show()
```

Box Plot

```python
Q1= df["total_bill"].quantile(0.25)

Q1 = df["total_bill"].quantile(0.25)

Q3 = df["total_bill"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Boundary:", lower)
print("Upper Boundary:", upper)
```
