```python
import seaborn as sns
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [35, 40, 50, 60, 75, 90]

sns.scatterplot(
    x=hours,
    y=marks,
    color="green",
    s=100
)

plt.title("Hours Studied vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```
